# Buy Crypto Press Release FAQ Generator: Search Walkthrough

This notebook runs the [Buy Crypto Press Release FAQ Generator](https://github.com/buycryptopressrelease/buy-crypto-pressrelease-faq-generator) on 50 real world questions about buying a crypto press release, checks each one finds the expected FAQ, lists the packages, and generates the FAQ with FAQPage schema. Source: [BuyCryptoPressRelease.com](https://buycryptopressrelease.com/).

## 1. Load the tool and the datasets

```python
import os, sys
import pandas as pd

BASES = ["/kaggle/input/buy-crypto-pressrelease-faq-dataset", "..", "."]
base = next(b for b in BASES if os.path.exists(os.path.join(b, "buy_crypto_pr_faq.py")))
data = base if os.path.exists(os.path.join(base, "crypto_pr_faqs.csv")) else os.path.join(base, "dataset")
sys.path.insert(0, base)
import buy_crypto_pr_faq as bc

faqs = pd.read_csv(os.path.join(data, "crypto_pr_faqs.csv"))
bench = pd.read_csv(os.path.join(data, "crypto_pr_search_benchmarks.csv"))
packages = pd.read_csv(os.path.join(data, "crypto_pr_packages.csv"))
print(len(faqs), "FAQs |", len(bench), "benchmark questions |", len(packages), "packages")
```

## 2. FAQs per category

```python
faqs.groupby("category").size().rename("faqs").to_frame()
```

## 3. Packages

```python
packages
```

## 4. Run every benchmark question

```python
rows = []
for r in bench.itertuples():
    a = bc.ask(r.query)
    rows.append({"query": r.query, "found": a["id"], "expected": r.expected_id,
                 "confidence": a["confidence"], "band": a["band"], "pass": a["id"] == r.expected_id})
results = pd.DataFrame(rows)
print(f"{results['pass'].sum()}/{len(results)} benchmark questions matched")
results
```

```python
results.groupby("band").size().rename("questions").to_frame()
```

## 5. Ask your own question

```python
for r in bc.search("can i pay with usdt", 3):
    print(f"Q{r['id']} [{r['confidence']}] {r['question']}")

best = bc.ask("how much does a crypto press release cost")
print("\n", best["answer"])
```

## 6. Generate the FAQ with FAQPage schema

```python
print(bc.export("jsonld", "pricing")[:1200])
```
